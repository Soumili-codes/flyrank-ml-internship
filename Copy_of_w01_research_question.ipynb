```python
import pandas as pd

url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Total rows:", len(df))
print("Total columns:", len(df.columns))
print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Find columns related to clicks, traffic, and engagement
for col in df.columns:
    if any(word in col.lower() for word in ["click", "ctr", "engagement", "view", "traffic", "impression"]):
        print(col)
```

```python
print("Total pages:", len(df))
print("Total clicks in 90 days:", df["clicks_90d"].sum())
print("Total impressions in 90 days:", df["impressions_90d"].sum())

print("\nAverage CTR:", round(df["ctr"].mean(), 4))
print("Average engagement rate:", round(df["engagement_rate"].mean(), 4))
```
